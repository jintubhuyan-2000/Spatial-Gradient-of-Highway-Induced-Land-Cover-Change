```python
import numpy as np
```

```python
# File path
StudyArea = '/content/drive/MyDrive/TZPR_NLP_Research/TZPR_NLP_LULC_2025.tif'
```

```python
with rasterio.open(StudyArea) as src:
    lulc = src.read(1)
    profile = src.profile
    transform = src.transform
    crs = src.crs
    bounds = src.bounds

# Print information
print("Raster information")
print("-------------------")
print("Width:", profile['width'])
print("Height:", profile['height'])
print("CRS:", crs)
print("Resolution:", profile['transform'].a, "m")
print("Bounds:", bounds)
print("NoData:", profile['nodata'])
```

```python
print("\nUnique LULC classes:")
print(np.unique(lulc))
```

```python
class_names={0:'water',1:'trees',2:'grass',3:'flooded vegetation',4: 'crops',5:'shrub and scrubs',6: 'built up',7:'bareland',8:'snow'}
```

```python
from matplotlib.colors import ListedColormap

# Dynamic World palette
dw_colors = [
    '#419BDF',  # Water
    '#397D49',  # Trees
    '#88B053',  # Grass
    '#7A87C6',  # Flooded vegetation
    '#E49635',  # Crops
    '#DFC35A',  # Shrub & scrub
    '#C4281B',  # Built
    '#A59B8F',  # Bare
    '#B39FE1'   # Snow & ice
]

cmap = ListedColormap(dw_colors)

plt.figure(figsize=(12, 10))

plt.imshow(
    lulc,
    cmap=cmap,
    vmin=0,
    vmax=8,
    interpolation='nearest'
)

plt.title(
    'Land Use/Land Cover (LULC), 2025',
    fontsize=17,
    fontweight='bold'
)

plt.xlabel('Pixel Column')
plt.ylabel('Pixel Row')

cbar = plt.colorbar(
    ticks=np.arange(9),
    fraction=0.035,
    pad=0.03
)

cbar.ax.set_yticklabels(
    list(class_names.values())
)

cbar.set_label(
    'Land-Cover Class',
    fontsize=12
)

plt.tight_layout()
plt.show()
```
